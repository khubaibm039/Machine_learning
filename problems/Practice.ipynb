Question: Using the provided data.csv file, write code to find:
1. Total missing values in each column
2. Percentage of missing values per column
3. Which columns have more than 50% missing values?


```python
total_missing_value = df.isnull().sum()
percentage_of_missing_values = df.isnull().mean() * 100
missing_upto_50 = percentage_of_missing_values[percentage_of_missing_values > 50]

print(total_missing_value.idxmax())
print(percentage_of_missing_values)
print(missing_upto_50)
print(total_missing_value)
```

Problem 3:
Question: Look at the Temperature column in data.csv. It contains:
1. Numeric values like 22.5, 18.6
2. String 'Unknown'
3. NaN values 

Write code to handle this column properly before imputation.



```python
from sklearn.impute import SimpleImputer

df['Temperature'] = df['Temperature'].replace('Unknown', np.nan)

temp_imputer = SimpleImputer(missing_values=np.nan, strategy='median')
df['Temperature'] = temp_imputer.fit_transform(df[['Temperature']]).ravel()
print(df['Temperature'].isnull().sum())
df['Temperature'].describe()

```

Problem 4:
Question: For the Diagnosis column in data.csv, implement imputation with:
1. Constant strategy using 'Missing' string
2. Add a missing indicator column

Write the complete code using SimpleImputer.


```python
from sklearn.impute import SimpleImputer

diag_imputer = SimpleImputer(missing_values=np.nan, strategy='constant', fill_value='Missing', add_indicator=True)
df[['Diagnosis', 'Diag_missing_indicator']] = diag_imputer.fit_transform(df[['Diagnosis']])
```
