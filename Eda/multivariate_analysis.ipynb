```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../Files/titanic_data_updated.csv')
df.head()
```

```python
sns.countplot(x=df['Sex'], hue=df['Survived'])
```

```python
gender_split = df.groupby('Sex')['Survived'].value_counts(normalize=True)
print(gender_split)
```

```python
sns.countplot(x=df['Pclass'], hue=df['Survived'])
```

```python
pclass_split = df.groupby('Pclass')['Survived'].value_counts(normalize=True)

print(pclass_split)
```

```python
sns.countplot(x=df['Embarked'], hue=df['Survived'])
```

```python
Embarked_split = df.groupby('Embarked')['Survived'].value_counts(normalize=True)

print(Embarked_split)
```

```python
sns.barplot(x=df['Pclass'], y=df['Fare'])
```

```python
sns.barplot(x=df['Survived'], y=df['Age'])
```

```python
sns.barplot(x=df['Survived'], y=df['Age'], hue=df['Sex'])
```

```python
sns.kdeplot(data=df, x='Age')
```

```python
sns.kdeplot(data=df, x='Age', hue='Survived')
```

```python
sns.kdeplot(data=df, x='Age', hue='Sex')
```

```python
sns.kdeplot(data=df, x='Age', hue='Pclass')
```
